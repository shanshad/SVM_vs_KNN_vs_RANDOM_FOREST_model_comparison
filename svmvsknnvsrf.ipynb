Goal : Evaluate and compare the performance of three classification algorithms—k-Nearest Neighbors (kNN) with n_neighbors ∈ {3, 5, 7}, Support Vector Machine (SVM) with RBF kernel regularization parameter C ∈ {0.1, 1, 10}, and Random Forest with n_estimators ∈ {100, 200} and max_depth ∈ {3, 5, None} — using the Breast Cancer Wisconsin dataset available in scikit-learn. Split it into training and test sets (75 % train / 25 % test) using train_test_split with random_state=42 and stratify=y. Standardize the features using StandardScaler. Use the same train and test sets for the classifiers. Submit a single ipynb file with all the three classifiers.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,f1_score,precision_score,recall_score
```

Here we have imported SVC,KNeighborsClassifier,RandomForestClassifier from sklearn library. for comparison between models and effect of changing hyper parameters we will be using accuracy score,f1 score,precision score and recall score. for spliting the data train_test_split and for standardizing features StandardScaler will be used

```python
from sklearn.datasets import load_breast_cancer
df=load_breast_cancer()
X,y,features=df.data,df.target,df.feature_names
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42,stratify=y)
print("before spliting the size of features ",X.shape)
print("after splitting the size of features ",X_train.shape)
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
x_vis_ns=pd.DataFrame(X)
y_vis=pd.DataFrame(y)
x_vis_s=pd.DataFrame(X_train_scaled)
print("features before scaling \n",x_vis_ns.head())
print("features after scaling \n",x_vis_s.head())
print("target column sample \n",y_vis.head())
```

Here we have imported our data set, splitted it for training and testing in the ratio of 75:25. then we have standardised the data and presented it to show how the data is changed after scaling. and the target is given in forms of 0 and 1.

Since we have splitted and standerdised our data we will start to train our models with this data.

**1. SVM Model**

```python
def svm_model(X_train,X_test,y_train,y_test):
    c=[0.1,1,10]
    acc=[]
    prec=[]
    recall=[]
    f1=[]
    for i in range(len(c)):
        svm_clf=SVC(kernel='rbf',C=c[i],gamma='scale')
        svm_clf.fit(X_train,y_train)
        y_pred=svm_clf.predict(X_test)
        accuracy=accuracy_score(y_test,y_pred)
        acc.append({"C value" : c[i],"accuracy":accuracy})
        precision=precision_score(y_test,y_pred)
        prec.append({"C value" : c[i],"precision":precision})
        f1_sc=f1_score(y_test,y_pred)
        f1.append({"C value" : c[i],"f1 score":f1_sc})
        recall_sc=recall_score(y_test,y_pred)
        recall.append({"C value" : c[i],"recall score":recall_sc})
    return pd.DataFrame(acc),pd.DataFrame(prec),pd.DataFrame(recall),pd.DataFrame(f1)
acc_svm,prec_svm,recall_svm,f1_svm=svm_model(X_train_scaled,X_test_scaled,y_train,y_test)
print(acc_svm,'\n',prec_svm,'\n',recall_svm,'\n',f1_svm)
```

Here we can see how the metrics are changing with C value

**2. KNN**

```python
def knn_model(X_train,X_test,y_train,y_test):
    n=[3,5,7]
    acc=[]
    prec=[]
    recall=[]
    f1=[]
    for i in range(len(n)):
        knn_model=KNeighborsClassifier(n_neighbors=n[i])
        knn_model.fit(X_train,y_train)
        y_pred=knn_model.predict(X_test)
        accuracy=accuracy_score(y_test,y_pred)
        acc.append({"n_neighbors":n[i],"accuracy":accuracy})
        precision=precision_score(y_test,y_pred)
        prec.append({"n_neighbors":n[i],"precision":precision})
        f1_sc=f1_score(y_test,y_pred)
        f1.append({"n_neighbors":n[i],"f1 score":f1_sc})
        recall_sc=recall_score(y_test,y_pred)
        recall.append({"n_neighbors":n[i],"recall score":recall_sc})
    return pd.DataFrame(acc),pd.DataFrame(prec),pd.DataFrame(recall),pd.DataFrame(f1)

acc_knn,prec_knn,recall_knn,f1_knn=knn_model(X_train_scaled,X_test_scaled,y_train,y_test)
print(acc_knn,'\n',prec_knn,'\n',recall_knn,'\n',f1_knn)
```

Here we can see how the metrics are changing with n neighbors value

**3.Random Forest**

```python
def random_forest_model(X_train,X_test,y_train,y_test):
    n=[100,200]
    max_dep=[3,5,None]
    acc=[]
    prec=[]
    recall=[]
    f1=[]
    for i in range(len(max_dep)):
        for j in range(len(n)):
            rf_model=RandomForestClassifier(n_estimators=n[j],max_depth=max_dep[i])
            rf_model.fit(X_train,y_train)
            y_pred=rf_model.predict(X_test)
            accuracy=accuracy_score(y_test,y_pred)
            acc.append({"max depth":max_dep[i],"n estimator ": n[j],"accuracy" : accuracy})
            precision=precision_score(y_test,y_pred)
            prec.append({"max depth":max_dep[i],"n estimator ": n[j],"precision":precision})
            f1_sc=f1_score(y_test,y_pred)
            f1.append({"max depth":max_dep[i],"n estimator ": n[j],"f1 score":f1_sc})
            recall_sc=recall_score(y_test,y_pred)
            recall.append({"max depth":max_dep[i],"n estimator ": n[j],"recall score":recall_sc})
    return pd.DataFrame(acc),pd.DataFrame(prec),pd.DataFrame(recall),pd.DataFrame(f1)
acc_rf,prec_rf,recall_rf,f1_rf=random_forest_model(X_train_scaled,X_test_scaled,y_train,y_test)
print(acc_rf,'\n',prec_rf,'\n',recall_rf,'\n',f1_rf)
```

Here we can see how metrics are changing with respect to max depth and n estimators

Now combine all data in to a single table for comparison

```python
svm_merged=acc_svm.merge(prec_svm,on='C value').merge(recall_svm,on='C value').merge(f1_svm,on='C value')
svm_merged.index=['svm(C=0.1)', 'svm(C=1)', 'svm(C=10)']
svm_merged=svm_merged.drop(columns='C value')
knn_merged=acc_knn.merge(prec_knn,on='n_neighbors').merge(recall_knn,on='n_neighbors').merge(f1_knn,on='n_neighbors')
knn_merged.index=['knn(n_neighbor=3)','knn(n_neighbors=5)','knn(n_neighbors=7)']
knn_merged=knn_merged.drop(columns='n_neighbors')
rf_merged=acc_rf.merge(prec_rf,on=['max depth','n estimator ']).merge(recall_rf,on=['max depth','n estimator ']).merge(f1_rf,on=['max depth','n estimator '])
rf_merged['(max depth, n estimator)'] = list(zip(rf_merged['max depth'], rf_merged['n estimator ']))
rf_merged=rf_merged.drop(columns=['max depth','n estimator '])
rf_merged=rf_merged[['(max depth, n estimator)','accuracy','precision','recall score','f1 score']]
idx=rf_merged['(max depth, n estimator)']
rf_merged.index=[f'rf(max depth, n estimator = {i})' for i in idx]
rf_merged=rf_merged.drop(columns=['(max depth, n estimator)'])
total=pd.concat([svm_merged,knn_merged,rf_merged])
print(total)
```

Comaring models for max accuracy,precision,recall score and f1 score

```python
max_accuracy=total['accuracy'].max()
max_accuracy_model=total['accuracy'].idxmax()
max_precision=total['precision'].max()
max_precision_model=total['precision'].idxmax()
max_recall=total['recall score'].max()
max_recall_model=total['recall score'].idxmax()
max_f1=total['f1 score'].max()
max_f1_model=total['f1 score'].idxmax()

print(f'the maximum accuracy = {max_accuracy} and the model = {max_accuracy_model}')
print(f'the maximum precision = {max_precision} and the model = {max_precision_model}')
print(f'the maximum recall score = {max_recall} and the model = {max_recall_model}')
print(f'the maximum f1 score = {max_f1} and the model = {max_f1_model}')
```
